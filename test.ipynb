```python
import nltk
from nltk.corpus import stopwords
import pickle
```

```python
def custom_tokenizer(text):
    tokens = text.split()
    return tokens
```

```python
with open('count_vectorizer.pkl','rb') as file:
    cv=pickle.load(file)

with open('rough_classifier_svm.pkl','rb') as file:
    clf=pickle.load(file)
```

```python
stop=stopwords.words('nepali')
```

```python
def remove_stopwords(text):
    tokens=text.split()
    cleaned_text=[word for word in tokens if word not in stop]
    return " ".join(cleaned_text)
```

```python
def query_creator(text):
    text=remove_stopwords(text)
    print(text)
    q_bow=cv.transform([text]).toarray()
    return q_bow
```

```python
q_bow=query_creator("जाठा माया गर्दा रैछन")
result=clf.predict(q_bow)
if result==1:
    print('rough sentence')
else:
    print('Non-rough sentence')
```

```python
q_bow=query_creator("देश राम्रो छ")
result=clf.predict(q_bow)
if result==1:
    print('rough sentence')
else:
    print('Non-rough sentence')

q_bow=query_creator(" माचिक्ने देश राम्रो छ")
result=clf.predict(q_bow)
if result==1:
    print('rough sentence')
else:
    print('Non-rough sentence')
```
